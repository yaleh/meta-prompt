```python
from typing import TypedDict

class DemoDict(TypedDict):
    name: str
    age: int

demo_dict = DemoDict(name="John", age=30)

# Convert a dict into the class
demo_dict_from_dict = DemoDict({"name": "Jane", "age": 25})

print(demo_dict)
print(demo_dict_from_dict)
```

```python
class Person(TypedDict):
    name: str
    age: int

class Employee(Person):
    employee_id: int

class Manager(Employee):
    team_size: int

person = Person(name="John", age=30)
employee = Employee(name="Jane", age=25, employee_id=123)
manager = Manager(name="Bob", age=40, employee_id=456, team_size=10)

print(person)
print(employee)
print(manager)

# Convert objects between them
person_from_employee = Person(employee)
employee_from_manager = Employee(manager)

print(person_from_employee)
print(employee_from_manager)
```

```python
# Create two classes with the same keys, and convert objects between them
class Person2(TypedDict):
    name: str
    age: int

class Employee2(TypedDict):
    name: str

person2 = Person2(name="John", age=30)
employee2 = Employee2(name="Jane")

print(person2)
print(employee2)

employee2_from_person2 = Employee2(person2)
person2_from_employee2 = Person2(employee2)

print(employee2_from_person2)
print(person2_from_employee2)

print(type(person2))
print(type(employee2))
print(type(employee2_from_person2))
print(type(person2_from_employee2))


```

```python
from typing import TypedDict, List, Annotated

class Student(TypedDict):
    name: str
    age: int
    courses: List[str]
    grades: List[float]
    favorite_course_index: int

    def dump(self):
        # return a dict with all attrs
        return {
            'name': self.get('name'),
            'age': self.get('age'),
            'courses': self.get('courses'),
            'grades': self.get('grades'),
            'favorite_course_index': self.get('favorite_course_index')
        }


# Create a Student instance
alice = Student(
    name="Alice",
    age=20,
    courses=["Math", "Physics", "Computer Science"],
    grades=[85.5, 92.0, 88.5],
    favorite_course_index=1
)

# Print the student information
print("Student Information:")
print(f"Name: {alice['name']}")
print(f"Age: {alice['age']}")
print(f"Courses: {', '.join(alice['courses'])}")
print(f"Grades: {', '.join(map(str, alice['grades']))}")

# print the dump method output
print("Dump method output:")
print(alice.dump())

# Calculate and print the average grade
average_grade = sum(alice['grades']) / len(alice['grades'])
print(f"\nAverage Grade: {average_grade:.2f}")

# Add a new course and grade
alice['courses'].append("Chemistry")
alice['grades'].append(90.0)

print("\nUpdated Student Information:")
print(f"Courses: {', '.join(alice['courses'])}")
print(f"Grades: {', '.join(map(str, alice['grades']))}")

# Update favorite course
alice['favorite_course_index'] = 3

# Show that TypedDict is still a regular dict
print("\nKeys in Student TypedDict:")
for key in alice.keys():
    print(f"- {key}")
```
